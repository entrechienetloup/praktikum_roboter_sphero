```python
import numpy as np
import matplotlib.pyplot as plt

# x1-Wertebereich
x = np.linspace(0, 6, 200)

# Restriktionen nach x2 umgestellt
y1 = (x + 6) / 2  # -x1 + 2x2 <= 6
y2 = 5 - x        # x1 + x2 <= 5

plt.figure(figsize=(6, 6))

# Begrenzungsgeraden
plt.plot(x, y1, label=r'$-x_1 + 2x_2 = 6$')
plt.plot(x, y2, label=r'$x_1 + x_2 = 5$')

# Zulässiger Bereich
y_fill = np.minimum(y1, y2)
plt.fill_between(x, 0, y_fill, where=(y_fill >= 0), color='gray', alpha=0.3, label='Zulässiger Bereich')

# Optimum (Schnittpunkt der Restriktionen)
opt_x, opt_y = 4/3, 11/3
plt.plot(opt_x, opt_y, 'ro', label=r'Optimum $v^*=(\frac{4}{3}, \frac{11}{3})$')

# Iso-Z-Gerade (Zielfunktion durch das Optimum)
# min z = x1 - 3x2 => x2 = x1/3 - z/3; mit z = -29/3 => x2 = x1/3 + 29/9
plt.plot(x, x/3 + 29/9, 'r--', label=r'Zielfunktion $z = -\frac{29}{3}$')

# Achsen & Formatierung
plt.xlim(0, 5)
plt.ylim(0, 5)
plt.xlabel(r'$x_1$')
plt.ylabel(r'$x_2$')
plt.legend()
plt.grid(True)
plt.show()
```
